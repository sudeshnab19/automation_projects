```python
import pandas as pd
import numpy as np
```

```python
dfb= pd.read_excel('bank statement.xlsx')
dfb
```

```python
dfb['Reference']=dfb['Bank Reference']
dfb
```

```python
dfg=pd.read_excel('general ledger.xlsx')
dfg
```

```python
dfg['Reference']=dfg['GL Reference']
dfg
```

# Keeping all the records from Bank statement and GL in 1 file using merge function and Reference column is the key

```python
df1=pd.merge(dfb,dfg,on='Reference',how='outer',suffixes=('_bank','_gl'))
df1
```

# Check whether Amount is same for the matched  reference nos.

```python
df1['Additional comments'] = df1.apply(
    lambda x: "Amount Matched" if x['Bank Amount (INR)'] == x['GL Amount (INR)'] else "Amount Not Matched",
    axis=1
)
```

```python
df1
```

# Find out the recurring patterns

```python
df1['Bank Description']=df1['Bank Description'].str.lower()
df1['GL Description']=df1['GL Description'].str.lower()
df1
```

```python
df1['Bank Description split']=df1['Bank Description'].str.split()
df1['GL Description split']=df1['GL Description'].str.split()
df1
```

```python
s = ['salary', 'bill', 'rent', 'monthly', 'quarterly', 'yearly']
match =[]
for i in range(len(df1)):
    words = df1.loc[i, 'Bank Description split']  # Get the list from the column

    # isinstance usage to avoid errors when iterating through data where some entries may not be lists 
    # e.g., could be NaN, None, or a string
    if isinstance(words, list):
        # Go through each word in that list
        for word in words:
            if word in s:
                print(words)
                break  # No need to check further words if one match is found
```

```python
s = ['salary', 'bill', 'rent', 'monthly', 'quarterly', 'yearly']

# Create an empty list to store results
matches = []

# Loop through each row
for i in range(len(df1)):
    words = df1.loc[i, 'Bank Description split']  # get the value from the column

    found = None  # default: no match
    if isinstance(words, list):  # only check if it's actually a list
        for word in words:
            if word in s:
                found = word     # store the matched word
                break            # stop after first match
    
    matches.append(found)  # add result (word or None) to the list

# Add as a new column
df1["Matched Word"] = matches
```
