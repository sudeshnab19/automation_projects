```python
import pandas as pd
import numpy as np
```

```python
A ='old_month.xlsx'
```

# Reading the file

```python
old=pd.read_excel(A)
old
```

# Delete the whitespaces and make object datatype lowercase in file

```python
for col in old.select_dtypes(include=[object]):
    old[col] = old[col].str.strip()
```

```python
for col in old.select_dtypes(include=[object]):
    old[col] = old[col].str.lower()
old
```

import xlsxwriter

with pd.ExcelWriter('Sorted.xlsx',engine='xlsxwriter') as writer:
    old.to_excel(writer,sheet_name='nowhitespace',index=False)
print("Downloaded")

# Delete the duplicate entries from the file

```python
old=old.drop_duplicates()
```

# Adding extra column 

```python
old['source']='old file'
```

```python
old
```

# cleansing the new file with whitespaces deletion, lowercase and duplicate removal and adding new column

```python
B ='current_month.xlsx'
new=pd.read_excel(B)
for col in new.select_dtypes(include=[object]):
    new[col] = new[col].str.strip()
for col in new.select_dtypes(include=[object]):
    new[col] = new[col].str.lower()
new=new.drop_duplicates()
new['source']='new file'
new
```

# Select few columns for concat


```python
cols=['Imp','Fund_Imp','Fund_Dept_ID','Approver_ID','Proposed_Invoice_Approver','Title','Threshold_USD','source']
```

```python
combined=pd.concat([old[cols],new[cols]])
combined
```

# Matched entries in both the files with source file name

```python
Values_found_in_both=combined[combined.duplicated(subset=[col for col in combined.columns if col != 'source'],keep=False)] # False is used to keep matched records in both files
Values_found_in_both
```
