# In pivot, we can't aggregate or summarize the dataframe like pivot table

```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('weather.csv')
df
```

```python
new_pivot=df.pivot(index='date',columns='city')
new_pivot
```

```python
# Save the pivot table to an Excel file
with pd.ExcelWriter('pivot_creation.xlsx', engine='xlsxwriter') as writer:
    new_pivot.to_excel(writer, sheet_name='pivot1')
print('File downloaded')
```
