Pivot table- used to summarize and aggregate data inside dataframe
Pivot can't do that. 
To get the Grand total use and Subtotal, Margins=True

```python
table = pd.pivot_table(df, values='Product Bookings Net',columns='Fiscal Quarter ID',
                       index='Business Model', aggfunc='sum',margins=True)
table
```

```python
# Save the pivot table to an Excel file
with pd.ExcelWriter('pivot_creation2.xlsx', engine='xlsxwriter') as writer:
    table.to_excel(writer, sheet_name='pivot1')
```
