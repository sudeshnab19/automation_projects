```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_excel('Sample1.xlsx')
df
```

To show Grandtotals and Subtotals, using margins=True

```python
p=pd.pivot_table(df,index='Business Model',columns=['Fiscal Period ID','Fiscal Quarter ID'],values='Product Bookings Net',aggfunc='sum',margins=True)
```

```python
p
```

# Save the pivot table to an excel file

```python
with pd.ExcelWriter('newpivotbookings.xlsx',engine='xlsxwriter') as writer:
    p.to_excel(writer,sheet_name='pivot1')
print('File downloaded')
```
