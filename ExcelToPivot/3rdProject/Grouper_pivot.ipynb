# Group by month from date column use grouper. 
# Margin=True for GrandTotal not rquired in grouper as it already grouping as per months
Grouper is used in pandas and it needs 2 arguments- frequency i.e. occurence and key i.e. column

```python
df.pivot_table(index=pd.Grouper(freq='M',key='date'),columns='city')
```
